```python
def f(x):
    return (x - 3) ** 2

def grad_f(x):
    return 2 * (x - 3)

# Parameters
x = 0.0           # initial guess
learning_rate = 0.1
num_iterations = 2

print(f"Initial x: {x}, f(x): {f(x)}")

for i in range(num_iterations):
    gradient = grad_f(x)
    x = x - learning_rate * gradient
    print(f"Iteration {i+1}: x = {x:.4f}, f(x) = {f(x):.4f}")

print(f"\nOptimized x: {x:.4f}, minimum value of f(x): {f(x):.4f}")
```
