```python
import numpy as np

def f(x):
    """Function to minimize: f(x) = x^2 + 4x + 4"""
    return x**2 + 4*x + 4

def grad_f(x):
    """Gradient of f(x): f'(x) = 2x + 4"""
    return 2*x + 4

def gradient_descent(initial_x, tol=1e-6, max_iter=1000, alpha=1.0, beta=0.8):
    """
    Gradient descent with adaptive step size using backtracking line search.

    Parameters:
        initial_x : float — starting point
        tol       : float — tolerance for convergence
        max_iter  : int — maximum iterations
        alpha     : float — initial step size
        beta      : float — step size reduction factor (0 < beta < 1)
    """
    x = initial_x
    for i in range(max_iter):
        grad = grad_f(x)
        step_size = alpha

        # Backtracking line search: decrease step size until it reduces f(x)
        while f(x - step_size * grad) > f(x) - 0.5 * step_size * grad**2:
            step_size *= beta

        # Update x
        new_x = x - step_size * grad

        # Check for convergence
        if abs(new_x - x) < tol:
            print(f"Converged after {i+1} iterations.")
            break

        x = new_x

        print(f"Iter {i+1:3d}: x = {x:.6f}, f(x) = {f(x):.6f}, step_size = {step_size:.6f}")

    return x

# Run the algorithm
min_x = gradient_descent(initial_x=10.0)
print(f"\nMinimum found at x = {min_x:.6f}, f(x) = {f(min_x):.6f}")
```
