```python
import sympy as sp

# Define the variables
x, y = sp.symbols('x y')

# Define the function
f = x**2 + y**2 - 4*x - 6*y + 13

# Compute first-order partial derivatives
f_x = sp.diff(f, x)
f_y = sp.diff(f, y)

# Solve equations f_x = 0 and f_y = 0 to find critical points
critical_points = sp.solve([f_x, f_y], (x, y))
print("Critical point(s):", critical_points)

# Compute second-order partial derivatives
f_xx = sp.diff(f_x, x)
f_yy = sp.diff(f_y, y)
f_xy = sp.diff(f_x, y)

# Second derivative test (Hessian determinant)
D = f_xx * f_yy - f_xy**2
print("Second partial derivatives:")
print(f" f_xx = {f_xx}, f_yy = {f_yy}, f_xy = {f_xy}")
print("Discriminant (Hessian):", D)

# Evaluate at critical point(s)
# critical_points is a dictionary, so we need to access its items
for point in critical_points.items():
    point_dict = {point[0]: point[1]}
    d_val = D.subs(point_dict)
    fxx_val = f_xx.subs(point_dict)
    value = f.subs(point_dict)

    print(f"\nAt point {point_dict}:")
    print(f" Hessian D = {d_val}")
    print(f" f_xx = {fxx_val}")
    print(f" Function value = {value}")

    if d_val > 0:
        if fxx_val > 0:
            print(" → Local minimum")
        else:
            print(" → Local maximum")
    elif d_val < 0:
        print(" → Saddle point")
    else:
        print(" → Test inconclusive")
```
