```python
import random

# Graph distances
graph = {
    'A': {'B': 2, 'C': 3},
    'B': {'A': 2, 'C': 1, 'D': 4},
    'C': {'A': 3, 'B': 1, 'D': 2},
    'D': {'B': 4, 'C': 2}
}

eta = {}
tau = {}
for i in graph:
    eta[i] = {}
    tau[i] = {}
    for j in graph[i]:
        eta[i][j] = 1 / graph[i][j]
        tau[i][j] = 1.0

alpha = 1
beta = 2
rho = 0.5
Q = 1
n_ants = 3
n_iter = 10
start_node = 'A'
goal_node = 'D'


def choose_next_node(current, visited):
    neighbors = {k: v for k, v in graph[current].items() if k not in visited}
    if not neighbors:
        return None
    scores = {}
    total = 0
    for neighbor in neighbors:
        score = (tau[current][neighbor] ** alpha) * (eta[current][neighbor] ** beta)
        scores[neighbor] = score
        total += score
    probs = {k: v / total for k, v in scores.items()}
    r = random.random()
    cumulative = 0
    for node, prob in probs.items():
        cumulative += prob
        if r <= cumulative:
            return node
    return random.choice(list(probs.keys()))

best_route = None
best_length = float('inf')

for iteration in range(n_iter):
    all_routes = []
    for ant in range(n_ants):
        visited = [start_node]
        current = start_node
        total_length = 0
        while current != goal_node:
            next_node = choose_next_node(current, visited)
            if not next_node:
                total_length = float('inf')
                break
            total_length += graph[current][next_node]
            visited.append(next_node)
            current = next_node
        if total_length < float('inf'):
            all_routes.append((visited, total_length))
            if total_length < best_length:
                best_length = total_length
                best_route = visited

    # Evaporate pheromone
    for i in tau:
        for j in tau[i]:
            tau[i][j] *= (1 - rho)

    # Deposit pheromone
    for route, length in all_routes:
        deposit = Q / length
        for i in range(len(route) - 1):
            u, v = route[i], route[i + 1]
            tau[u][v] += deposit
            tau[v][u] += deposit  # symmetric graph

    print(f"Iteration {iteration+1}: Best route so far = {best_route} with length {best_length}")

print("\nFinal Best Route:", best_route)
print("Final Best Length:", best_length)
```
